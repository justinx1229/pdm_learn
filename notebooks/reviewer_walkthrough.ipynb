## Run the workflows

From the repository root, run:

```bash
uv run python scripts/reproduce.py all --data-dir /path/to/data --output-dir runs/my-analysis
```

For a small example, use `--data-dir tests/fixtures/reproduction --config tests/fixtures/reproduction/workflows.json --output-dir runs/example`. See `REPRODUCE.md` for the input layout and settings.

````python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'REPRODUCE.md').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
RUN_DIR = PROJECT_ROOT / 'runs' / 'my-analysis'  # Set to your --output-dir.
if not RUN_DIR.exists():
    raise FileNotFoundError(f'No outputs found in {RUN_DIR}. Run scripts/reproduce.py first.')
````

## Oncogene results

The ranking is generated from the PDM feature table. Method benchmarks are optional and require explicit negative labels when running the prediction script.

````python
oncogene_ranking = pd.read_csv(RUN_DIR / 'oncogene' / 'results' / 'oncogene_ranking.csv')
display(oncogene_ranking.head(10))
benchmark_path = RUN_DIR / 'oncogene' / 'results' / 'oncogene_method_benchmarks.csv'
if benchmark_path.exists():
    display(pd.read_csv(benchmark_path))
````

## PPI results

BioGRID pairs are represented with order-independent `canonical_pair` identifiers. Curated cancer-complex pairs define the positive labels.

````python
ppi_ranking = pd.read_csv(RUN_DIR / 'ppi' / 'results' / 'ranked_PPI_pairs_cancer.csv')
ppi_benchmark = pd.read_csv(RUN_DIR / 'ppi' / 'results' / 'ppi_control_benchmarks.csv')
display(ppi_ranking.head(10))
display(ppi_benchmark)
````
